```julia
using Pkg
using Dynare
using Serialization
using DataFrames
using XLSX
using Plots
```

```julia
m = @dynare("RBC_fiscal_estimacion.mod");
```

```julia
irfs_table = m.results.model_results[1].irfs[Symbol("Choque de PTF")]
irfs_df = DataFrame(irfs_table)
```

```julia
forecast_table = collect(values(m.results.model_results[1].forecast))[1]
forecast_df = DataFrame(forecast_table)
```

```julia
sm = m.results.model_results[1].smoother
smooth_df = DataFrame(sm)
```

```julia
function clear_sheet(sheet)
    num_rows = 500                      # Número arbitrário de linhas
    num_cols = 100                      # Número arbitrário de colunas
    
    for row in 1:num_rows
        for col in 1:num_cols
            sheet[row, col] = ""
        end
    end
end
```

```julia
XLSX.openxlsx("Forecast_results.xlsx", mode="rw") do xf
    sheet = xf["Forecast"]                                        # Selecionar a folha a modificar
    clear_sheet(sheet)                                            # Eliminar o conteúdo da folha
    
    # Escrever os nomes das colunas
    for (i, col_name) in enumerate(names(irfs_df))
        sheet[1, i] = col_name
    end
    # Escribir los datos del DataFrame
    for (i, row) in enumerate(eachrow(forecast_df))
        for (j, cell) in enumerate(row)
            sheet[i + 1, j] = cell
        end
    end
end
```

```julia
XLSX.openxlsx("Forecast_results.xlsx", mode="rw") do xf
    sheet = xf["Smooth"]                                          # Selecionar a folha a modificar
    clear_sheet(sheet)                                            # Eliminar o conteúdo da folha
    
    # Escrever os nomes das colunas
    for (i, col_name) in enumerate(names(irfs_df))
        sheet[1, i] = col_name
    end
    # Escrever os dados do DataFrame
    for (i, row) in enumerate(eachrow(smooth_df))
        for (j, cell) in enumerate(row)
            sheet[i + 1, j] = cell
        end
    end
end
```
