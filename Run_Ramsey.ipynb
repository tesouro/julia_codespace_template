```julia
m = @dynare("Ramsey.mod")
```

```julia
irfdata = m.results.model_results[1].irfs[:e_A]
```

```julia
vars = [:A, :y, :c, :w, :k, :r]
titles = ["A", "y", "c", "w", "k", "r"]

p = plot(layout = (3, 2), size = (1000, 600))

for (i, v) in enumerate(vars)
    plot!(p[i], irfdata[:, v],
          color = :black,
          title = titles[i],
          ylabel = "",
          legend = false)
    hline!(p[i], [0], color = :red, lw = 1, linestyle = :solid)
end

display(p)
savefig("Ramsey_IRFs.png") 
```

# Workshop

#### b) Suponha que a taxa de depreciação do capital aumenta.

```julia
mb = @dynare("Ramsey_b.mod")
```

```julia
irfdatab = mb.results.model_results[1].irfs[:e_A]
```

```julia
vars = [:A, :y, :c, :w, :k, :r]
titles = ["A", "y", "c", "w", "k", "r"]

p = plot(layout = (3, 2), size = (1000, 600))

for (i, v) in enumerate(vars)
    plot!(p[i], vec(irfdata[:, v]),
          color = :black,
          label = "delta = 0.015",
          title = titles[i])
    plot!(p[i], vec(irfdatab[:, v]),
          color = :blue,
          linestyle = :dash,
          label = "delta = 0.02")
    hline!(p[i], [0], color = :red, lw = 1, linestyle = :solid, label = "")
end

display(p)
savefig("Ramsey_IRFs_delta.png")
```

#### c) Ajuste o modelo para simular um choque de preferências.

```julia
mc = @dynare("Ramsey_c.mod")
```

```julia
irfdatac = mc.results.model_results[1].irfs[:e_beta]
```

```julia
vars = [:A, :y, :c, :w, :k, :r]
titles = ["A & β", "y", "c", "w", "k", "r"]

p = plot(layout = (3, 2), size = (1000, 600))

for (i, v) in enumerate(vars)
    irf_prod = vec(irfdata[:, v])
    if v == :A
        irf_pref = vec(irfdatac[:, :beta])
    else
        irf_pref = vec(irfdatac[:, v])
    end
    plot!(p[i], irf_prod,
          color = :black,
          label = "Choque de produtividade",
          title = titles[i],
          lw = 1)
    plot!(p[i], irf_pref,
          color = :green,
          linestyle = :dash,
          label = "Choque de preferências",
          lw = 1)
    hline!(p[i], [0], color = :red, lw = 1, linestyle = :solid, label = "")
end

display(p)
savefig("Ramsey_IRFs_beta.png")
```
